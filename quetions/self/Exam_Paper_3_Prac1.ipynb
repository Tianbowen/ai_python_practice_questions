```python
from tensorflow.keras.datasets import mnist

# 加载 mnist数据集
# 训练集图像数据 训练集标签 测试集图像数据 测试集标签

(train_image_data, train_image_label), (test_image_data, test_image_label) = mnist.load_data()

# 查看数据形状
print("="*50)
print("MNIST 数据集基本信息")
print("="*50)
print(train_image_data.shape)
print(f"训练集图像数据：{train_image_data.shape}")
print(f"训练集标签:{train_image_label.shape}")
print(f"测试集图像数据：{test_image_data.shape}")
print(f"测试集标签：{test_image_label.shape}")
print(f"图片尺寸:{train_image_data.shape[1]} * {train_image_data.shape[2]}")
print(f"像素值范围：{train_image_data.min()} ~ {train_image_data.max()}")
print(f"标签范围：{train_image_label.min()} ~ {train_image_label.max()}")
print()
```

```python
from tensorflow.keras.utils import to_categorical

# 数据归一化 将像素值从0-255 缩放到 0-1
train_image_data_norm = train_image_data.astype('float32') / 255.0
test_image_data_norm = test_image_data.astype('float32') / 255.0
# print(train_image_data_norm)
# print(test_image_data_norm)

# 展平为 784 维向量
train_image_data_vec = train_image_data_norm.reshape(train_image_data_norm.shape[0], -1)
test_image_data_vec = test_image_data_norm.reshape(test_image_data_norm.shape[0], -1)

# 标签 ont-hot 编码
train_image_label_cat = to_categorical(train_image_label, 10)
test_image_label_cat = to_categorical(test_image_label, 10)
print('预处理后数据形状：')
print(f"训练集图像：{train_image_data_vec.shape}")
print(f"训练集标签：{train_image_label_cat.shape}")
print(f"测试集图像：{test_image_data_vec.shape}")
print(f"测试集标签：{test_image_label_cat.shape}")
print(f"像素值范围检验：{train_image_data_vec.min():.2f} ~ {train_image_data_vec.max():.2f}")
```

```python
"""
    神经网络模型构建与训练计划
    使用Keras框架构建一个全连接神经网络模型，用于手写数字识别
    模型至少包含输入层，隐藏层和输出层
    编写训练代码，设置合适的优化器，损失函数和评估指标
    训练完成后保存模型到本地
    绘制包含损失和准确率变化曲线
"""

# Sequential 模型类，它是 Keras中最简单的模型类型。适合大多数前馈神经网络
from tensorflow.keras.models import Sequential
# 导入 常用的网络层 Dense 全连接层， Flatten展平层
# 全连接层， 每个输入节点都与该层的所有输出节点相连
# 展平层，将多维输入（二维图像，三维特征图）压平 一维向量
from tensorflow.keras.layers import Dense, Flatten
# 导入 Adam 优化器
from tensorflow.keras.optimizers import Adam

import matplotlib.pyplot as plt

model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# 显示模型结构
model.summary()

# 编译模型
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    train_image_data_vec, train_image_label_cat,
    validation_data=(test_image_data_vec, test_image_label_cat),
    epochs=10,
    batch_size=128,
    verbose=1
)

test_loss, test_accuracy = model.evaluate(test_image_data_vec, test_image_label_cat)
print(f"测试集损失率：{test_loss}, 测试集准确率：{test_accuracy}")

# 保存模型
model.save('model_prac1.h5') # h5 为HDF5格式

# 绘制训练过程中的损失和准确率曲线
# print(history.history)
train_loss = history.history['loss']
val_loss = history.history['val_loss']
train_accuracy = history.history['accuracy']
val_accuracy = history.history['val_accuracy']
epochs_range = range(1, len(train_loss) + 1)

# 设置图表中文
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(12, 4))

# 损失曲线
plt.subplot(1, 2, 1)
plt.plot(epochs_range, train_loss, 'b-', label='训练损失')
plt.plot(epochs_range, val_loss, 'r-', label='验证损失')
plt.title('训练与验证损失曲线')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# 准确率曲线
plt.subplot(1, 2, 2)
plt.plot(epochs_range, train_accuracy, 'y-', label='训练准确率')
plt.plot(epochs_range, val_accuracy, 'r-', label='验证准确率')
plt.title('训练与验证准确率曲线')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('training_curves_prac1.png', dpi=150)
plt.show()
```

```python
from tensorflow.keras.models import load_model
import os
import numpy as np

local_model = None
if os.path.exists('model_prac1.h5'):
    local_model = load_model('model_prac1.h5')
    print('model_prac1.h5 加载成功。')

if local_model is not None:
    print("模型期望输入形状:", local_model.input_shape)

num_samples = 5
indices = np.random.choice(len(test_image_data_norm), num_samples, replace=False)
sample_images = test_image_data_norm[indices]
sample_labels = test_image_label[indices]

# 对选中图片进行预测
predictions = local_model.predict(sample_images.reshape(num_samples, 784), verbose=0)
print(predictions)
predicted_classes = np.argmax(predictions, axis=1)
predicted_probs = np.max(predictions, axis=1)

# 使用matplotlib.pyplot 可视化结果
plt.figure(figsize=(12, 6))
for i in range(num_samples):
    plt.subplot(2, 3, i + 1)
    plt.imshow(sample_images[i], cmap='gray')
    plt.title(f"真实标签:{sample_labels[i]}\n预测标签：{predicted_classes[i]}\n概率：{predicted_probs[i]:.2f}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```
